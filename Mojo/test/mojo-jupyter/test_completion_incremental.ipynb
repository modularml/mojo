```mojo
struct Animal:
    var name: String
    var legs: Int

    fn __init__(out self, name: String, legs: Int):
        self.name = name
        self.legs = legs
```

```mojo
fn describe(a: Animal) -> String:
    return a.name
```

```mojo
struct Vehicle:
    var make: String
    var wheels: Int

    fn __init__(out self, make: String, wheels: Int):
        self.make = make
        self.wheels = wheels
```

```mojo
%%test_code_completion
fn use_animal(a: Animal):
    return a. 
```
